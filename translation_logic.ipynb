## Library Import

```python
from deep_translator import GoogleTranslator
from gtts import gTTS
import IPython.display as ipd
```

## Supported Languages

```python
languages = GoogleTranslator().get_supported_languages(as_dict=True)
print(list(languages.items())[:10])
```

## Simple Translation Test

```python
input_text = "Hello, how are you?"
source_lang = "en"       # English
target_lang = "ur"       # Urdu

translated = GoogleTranslator(source=source_lang, target=target_lang).translate(input_text)
print("Original:", input_text)
print("Translated:", translated)
```

## Auto-Detect Source Language Test 

```python
input_text2 = "Bonjour tout le monde"
translated2 = GoogleTranslator(source="auto", target="en").translate(input_text2)
print("Original:", input_text2)
print("Translated:", translated2)
```

## Text-to-Speech Test

```python
tts = gTTS(text=translated, lang=target_lang)
tts.save("output_audio.mp3")
ipd.Audio("output_audio.mp3")
```

## Function for Reuse

```python
def translate_text(text, source, target):
    """Simple function: text translate karke return karta hai"""
    return GoogleTranslator(source=source, target=target).translate(text)

# Test
result = translate_text("Good morning", "en", "hi")
print(result)
```
